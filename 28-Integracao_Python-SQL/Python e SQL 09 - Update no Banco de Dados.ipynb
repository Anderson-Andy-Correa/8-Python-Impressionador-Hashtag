```python
import pyodbc
from pathlib import Path
import pandas as pd

dados_conexao = (
    'driver={SQLite3 ODBC Driver};'
    'host=localhost;'
    fr'database={Path("./Dados/chinook.db")};'
)
```

```python
conexao = pyodbc.connect(dados_conexao)
cursor = conexao.cursor()

try:
    cursor.execute('''
SELECT * FROM customers WHERE CustomerId = 1
''')
    clientes = cursor.fetchall()
    colunas = [col[0] for col in cursor.description]
    display(pd.DataFrame.from_records(clientes, columns=colunas))
finally:
    cursor.close()
    conexao.close()
```

```python
try:
    conexao = pyodbc.connect(dados_conexao)
    cursor = conexao.cursor()
    cursor.execute('''
UPDATE customers SET FirstName = 'Anderson', Email = 'anderson.correa@outlook.com' WHERE CustomerId = 1
''')
    conexao.commit()
finally:
    cursor.close()
    conexao.close()
```
